````python
import base64
from pathlib import Path
from IPython import display as ipythondisplay

def show_videos(video_path, prefix=""):
    """Display MP4 videos from a folder in the notebook."""
    html = []
    for mp4 in Path(video_path).glob(f"{prefix}*.mp4"):
        video_b64 = base64.b64encode(mp4.read_bytes()).decode('ascii')
        html.append(f'''<video alt="{mp4}" autoplay loop controls style="height: 400px;">
            <source src="data:video/mp4;base64,{video_b64}" type="video/mp4" />
        </video>''')
    ipythondisplay.display(ipythondisplay.HTML(data="<br>".join(html)))
````

````python
import gymnasium as gym
import numpy as np

class GoalCurriculumWrapper(gym.Wrapper):
    """
    Gymnasium-compatible curriculum wrapper for Fetch* goal environments.
    reset() -> (obs, info)
    step()  -> (obs, reward, terminated, truncated, info)
    """
    def __init__(
        self,
        env,
        start_radius=0.02,
        end_radius=0.15,
        total_steps=1_000_000,
        seed=0,
    ):
        super().__init__(env)
        self.start_radius = float(start_radius)
        self.end_radius = float(end_radius)
        self.total_steps = int(total_steps)
        self.steps = 0
        self.rng = np.random.default_rng(seed)

    def reset(self, **kwargs):
        obs, info = self.env.reset(**kwargs)

        # progress in [0, 1]
        p = min(1.0, self.steps / max(1, self.total_steps))
        radius = self.start_radius + p * (self.end_radius - self.start_radius)

        # base goal: prefer existing desired_goal, else achieved_goal
        base_goal = obs.get("desired_goal", obs["achieved_goal"]).copy()
        new_goal = base_goal.copy()

        # Push tasks: vary x/y on table, keep z
        new_goal[:2] += self.rng.uniform(-radius, radius, size=2)

        # Update env internal goal so reward matches the new goal
        if hasattr(self.env.unwrapped, "goal"):
            self.env.unwrapped.goal = new_goal

        obs["desired_goal"] = new_goal
        return obs, info

    def step(self, action):
        self.steps += 1
        return self.env.step(action)
````

````python
import numpy as np

def make_env():
    env = gym.make("FetchPushDense-v4")
    env = GoalCurriculumWrapper(env, total_steps=1_000_000, seed=0)
    return env


from stable_baselines3 import DDPG
from stable_baselines3.common.vec_env import DummyVecEnv, VecMonitor
from stable_baselines3.common.noise import NormalActionNoise


env = DummyVecEnv([make_env])
env = VecMonitor(env)

n_actions = env.action_space.shape[-1]
action_noise = NormalActionNoise(
    mean=np.zeros(n_actions),
    sigma=0.2 * np.ones(n_actions),
)

model = DDPG(
    policy="MultiInputPolicy",
    env=env,
    learning_rate=1e-3,
    batch_size=256,
    buffer_size=1_000_000,
    learning_starts=10_000,
    train_freq=1,
    gradient_steps=1,
    gamma=0.98,
    tau=0.02,
    action_noise=action_noise,
    policy_kwargs=dict(net_arch=[256, 256]),
    verbose=1,
)

model.learn(total_timesteps=1_000_000)
model.save("ddpg_fetchpush_curriculum")

env.close()
````
